```python
# Create a dictionary with 5 student details
my_dict = {
    "name": "Atharv Patil",
    "rollno": 1272261910,
    "division": 12,
    "branch": "CSE AI & DS",
    "age": 20
}

#Display  original dictionary
print("Original Student Information:")
print(my_dict)

#Add a new  student information
my_dict["city"] = "Pune"

print("\nAfter Adding City:")
print(my_dict)

#Delete student information
del my_dict["age"]

print("\nAfter Deleting Age:")
print(my_dict)

#Display final student information
print("\nFinal Student Information:")
for key, value in my_dict.items():
    print(key, ":", value)
```

```python
#Convert list to dictionary

list1 = ["name", "panel" , "rollno"]
list2 = ["ABC", "B" , "34"]

#Create dictionary
my_dict = dict(zip(list1, list2))

#Display dictionary
print("Dictionary is:")
print(my_dict)
```

```python
# Sort the elements of the dictionary using sorted()

my_dict = {
    "name": "ABC",
    "panel": "B",
    "rollno": 34,
    "marks": [65, 87, 67, 94]
}

# Sorting dictionary keys 
sorted_keys = sorted(my_dict)

print("Sorted Dictionary Keys:")
print(sorted_keys)
```

```python
# Convert dictionary to seperate list of keys and values

my_dict = {
    "name": "ABC",
    "panel": "B",
    "rollno": 34,
    "marks": [65, 87, 67, 94]
}

#Convert keys and values into lists
keys_list = list(my_dict.keys())
values_list = list(my_dict.values())

print("Keys List:", keys_list)
print("Values List:", values_list)
```

```python
# Find the mean of values in a dictionary

mydict = {
    "marks1": 23,
    "marks2": 123,
    "marks3": 43,
    "marks4": 13,
    "marks5": 39
}

#Calculate mean
mean = sum(mydict.values()) / len(mydict)

print("Mean of all values =", mean)
    
```

```python
# Calculating frequency of letters in a string

string = input("Enter a string:Python")

frequency = {}

for letter in string:
    if letter in frequency:
        frequency[letter] += 1
    else:
        frequency[letter] = 1

print("Frequency of letters:")
print(frequency)
```

```python
# Take numerator and denominator from the users 

numerator = int(input("Enter numerator: "))
denominator = int(input("Enter denominator: "))

# Find quotient and remainder
result = (numerator // denominator, numerator % denominator)

# Display tuple
print("Quotient and Remainder:", result)
```

```python
# Find number of days between two dates using tuples

from datetime import date 

# Input first date
d1 = int(input("Enter first date (DD): "))
m1 = int(input("Enter first month (MM): "))
y1 = int(input("Enter first year (YYYY): "))

# Input second date
d2 =  int(input("Enter second date (DD): "))
m2 =  int(input("Enter second month (MM): "))
y2 =  int(input("Enter second year (YYYY): "))

# Create tuples
date1 = (d1 , m1, y1)
date2 = (d2 , m2, y2)

# Convert tuples to date objects 
first_date = date(date1[2], date1[1], date1[0])
second_date = date(date2[2], date2[1], date2[0])

# Find difference 
days = abs((second_date - first_date).days)

print("Date 1:", date1)
print("Date 2:", date2)
print("Number of days between two dates:", days)
            
```
