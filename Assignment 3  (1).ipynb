```python
def prime_factors(n):
    print("Prime factors are:")
    result = []
    i = 2
    while i <= n:
        if n % i == 0:
            result.append(str(i))
            while n % i == 0:
                n = n // i
        i += 1
    print(",".join(result))

num = int(input("Enter a number: "))
prime_factors(num)
```

```python
def even_numbers():
    print("Even numbers from 1 to 100:")
    for i in range(2, 101, 2):
        print(i, end=" ")

even_numbers()
```

```python
import math

def quadratic_roots(a, b, c):
    try:
        discriminant = b**2 - 4*a*c
        
        # This line will create ValueError if discriminant < 0
        sqrt_d = math.sqrt(discriminant)
        
        root1 = (-b + sqrt_d) / (2*a)
        root2 = (-b - sqrt_d) / (2*a)
        
        if discriminant == 0:
            print(f"Real and same roots: {root1}")
        else:
            print(f"Real and different roots: {root1} and {root2}")

    except ValueError as e:
        print(f"ValueError: {e} - No real roots (b^2 - 4ac < 0)")
    except NameError as e:
        print(f"NameError: {e} - Variable not defined")
    except TypeError as e:
        print(f"TypeError: {e} - Please enter numeric coefficients only")
    except ZeroDivisionError:
        print("ZeroDivisionError: Coefficient 'a' cannot be 0")

# Input from user
try:
    a = float(input("Enter coefficient a: "))
    b = float(input("Enter coefficient b: "))
    c = float(input("Enter coefficient c: "))
    quadratic_roots(a, b, c)
except TypeError:
    print("Invalid input type")
```
